## Step 1 — Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print('Libraries loaded successfully!')
```

## Step 2 — Load the Dataset

```python
df = pd.read_csv('supermarket_sales.csv')

print('Shape:', df.shape)
df.head()
```

## Step 3 — Explore the Data

```python
print('Column names:')
print(df.columns.tolist())
print('\nData types:')
print(df.dtypes)
print('\nMissing values:')
print(df.isnull().sum())
```

```python
df.describe()
```

## Step 4 — Clean the Data

```python
# Convert Date to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Extract useful columns
df['Month'] = df['Date'].dt.month_name()
df['Hour'] = pd.to_datetime(df['Time'], format='%H:%M').dt.hour

print('New columns added: Month, Hour')
df[['Date', 'Month', 'Time', 'Hour']].head()
```

## Step 5 — Visualizations

```python
# Chart 1: Sales by Product Line
product_sales = df.groupby('Product line')['Total'].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(product_sales.index, product_sales.values, color='steelblue')
plt.title('Total Sales by Product Category', fontsize=14)
plt.xlabel('Product Category')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Revenue by City (Pie Chart)
city_sales = df.groupby('City')['Total'].sum()

plt.figure(figsize=(7, 7))
plt.pie(city_sales.values, labels=city_sales.index, autopct='%1.1f%%',
        colors=['#4C72B0', '#55A868', '#C44E52'])
plt.title('Revenue Share by City', fontsize=14)
plt.show()
```

```python
# Chart 3: Customer Visits by Hour
hourly = df['Hour'].value_counts().sort_index()

plt.figure(figsize=(10, 5))
plt.plot(hourly.index, hourly.values, marker='o', color='darkorange', linewidth=2)
plt.title('Customer Visits by Hour of Day', fontsize=14)
plt.xlabel('Hour (24h format)')
plt.ylabel('Number of Transactions')
plt.xticks(hourly.index)
plt.tight_layout()
plt.show()
```

```python
# Chart 4: Average Rating by Branch
branch_rating = df.groupby('Branch')['Rating'].mean().sort_values(ascending=False)

plt.figure(figsize=(7, 5))
bars = plt.bar(branch_rating.index, branch_rating.values,
               color=['#4C72B0', '#55A868', '#C44E52'])
for bar, val in zip(bars, branch_rating.values):
    plt.text(bar.get_x() + bar.get_width()/2, val + 0.1,
             f'{val:.2f}', ha='center', fontsize=12)
plt.title('Average Customer Rating by Branch', fontsize=14)
plt.xlabel('Branch')
plt.ylabel('Average Rating (out of 10)')
plt.ylim(0, 10)
plt.tight_layout()
plt.show()
```

## Step 6 — Summary of Findings

```python
print('=' * 50)
print('KEY INSIGHTS')
print('=' * 50)
print(f'Total Revenue:         ${df["Total"].sum():,.2f}')
print(f'Average Transaction:   ${df["Total"].mean():,.2f}')
print(f'Best Product Category: {product_sales.idxmax()}')
print(f'Top City by Revenue:   {city_sales.idxmax()}')
print(f'Peak Shopping Hour:    {hourly.idxmax()}:00')
print(f'Highest Rated Branch:  Branch {branch_rating.idxmax()}')
print(f'Most Used Payment:     {df["Payment"].value_counts().idxmax()}')
```
